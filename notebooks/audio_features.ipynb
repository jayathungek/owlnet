```python
%load_ext autoreload
%autoreload 2

import os
import sys
sys.path.append("/root/owlnet")
os.environ["CUDA_VISIBLE_DEVICES"] = "0, 1"

from pathlib import Path

import torch

from owlnet.core.utils import load_config, gather_data_files, chop_file
```

```python
data_dir = Path("owl_data/stoke_wake")
all_clips = data_dir.glob("*")
all_clips = [d for d in all_clips if d.is_dir()]
for d in all_clips:
    filename = d.name
    config = load_config(d / "config.json")
    file = Path(config["data_dir"]) / config["test_file"]
    chunks, original_chunks, chunk_crossing_times = chop_file(config, file, display=False)
    for i, chunk in enumerate(chunks):
        save_obj = {
            "chunk": chunk,
            "chunk_original": original_chunks[i],
            "chunk_crossing_times": chunk_crossing_times[i]
        }
        torch.save(save_obj, d / f"{filename}_{i}.pth")
```
