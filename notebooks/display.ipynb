```python
from owlnet.core.utils import load_config
from owlnet.data.dataloading import load_data


config = load_config(f"{root_dir}/settings/config.json")
load_obj = load_data(config)
```

```python
from owlnet.core.utils import get_model

owlnet = get_model(config, "nest0samples_100.epochs")
owlnet.eval()
pass
```
