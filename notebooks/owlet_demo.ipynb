```python
from owlnet.core.utils import get_model


owlnet = get_model(config)
owlnet.eval()
pass
```

```python
from ipywidgets import widgets

from owlnet.data.dataloading import CollateFunc, get_all_validation_embeds, create_embeds
from owlnet.core.interactive import VisualiserInteractive, ControlPanel


collate_func = CollateFunc(spec_height=config['spec_height'])
embeddings, melspecs, melspecs_og, _ = create_embeds(config, owlnet, owlet_train)
vis = VisualiserInteractive(config, embeddings, melspecs, melspecs_og)
vis.show()


all_val_embeds = get_all_validation_embeds(config, owlnet, owlet_dataset, collate_func)
cpanel = ControlPanel(config, all_val_embeds, vis)
display(widgets.VBox([cpanel.dataset_image, cpanel.progress, cpanel.progress_text]))
display(widgets.HBox([cpanel.step_button, cpanel.reset_button, cpanel.hop_size_buttons]))
```
